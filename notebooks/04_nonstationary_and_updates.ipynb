# Non-stationary kernels and two-timescale updates

This notebook mirrors `docs/05_non_stationary_gp_spectral.md` and `docs/06_two_timescale_bayesian_update.md`.

```python
import numpy as np
from gp_foundations.kernels import MaternKernel, TimeDecayKernel
from gp_foundations.updates import TwoTimescaleUpdater

kernel = TimeDecayKernel(MaternKernel(length_scale=0.2), decay=0.5, reference_point=0.0)
gram = kernel(np.linspace(0.0, 1.0, 4)[:, None])

updater = TwoTimescaleUpdater(
    fast_update=lambda state, obs, step: (state or 0) + obs,
    slow_update=lambda state, obs, step: state,
    slow_interval=3,
)
for obs in [1, 2, 3, 4]:
    updater.update(obs)
gram.shape, updater.fast_updates, updater.slow_updates
```
